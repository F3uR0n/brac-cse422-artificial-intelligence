```python
import heapq

grid= [
    ['D', 0, 0, 'X', 0],
    [0, 'X', 0, 0, 0],
    [0, 0, 'X', 0, 0],
    ['X',0, 0, 0, 0],
    [0,0, 0, 'X', 'P']
]
start = (0, 0)
goal = (4,4)

def heuristic(a,b):
  return abs( a[0]-b[0] ) + abs( a[1]-b[1] )

def aStarGrid(grid, start, goal):
  row = len(grid)
  col = len(grid[0])
  gcost = {start: 0}
  parent = {start: None}
  pq= []
  heapq.heappush(pq, (heuristic(start, goal), start))

  while pq:
    f, u = heapq.heappop(pq)
    totalCost = gcost[u]

    if u==goal:
      return totalCost
      # path=[]
      # while u!=None:
      #   path.append(u)
      #   u = parent[u]
      # return totalCost, path[::-1]

    x, y = u
    pos = [ (-1,0), (1,0), (0,1), (0,-1) ]
    for dx, dy in pos:
      nx, ny = dx+x, dy+y
      if row>nx>=0 and col>ny>=0 and grid[nx][ny]!='X':
        newG = gcost[u] + 1
        newF = newG + heuristic( (nx, ny), goal )

        if newG not in gcost or newG<gcost[(nx, ny)]:
          gcost[(nx, ny)] = newG
          # parent[(nx, ny)] = u
          heapq.heappush(pq, (newF, (nx, ny)))

  return 'No Path'

print(aStarGrid(grid, start, goal))
```
